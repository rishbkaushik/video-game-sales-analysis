# 6. GENRE WISE DISTRIBUTION


```python
a=df['Genre'].value_counts() 
plt.figure(figsize=(15, 5))
plt.bar(a.index,a)
plt.show()
```

# 7. GLOBAL SALES BY EACH PLATFORM

```python
b=df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False)
print("Platform with the highest total global sales:")
print(b.head(1))
plt.figure(figsize=(15, 5))
plt.bar(b.index,b)
plt.title('Total Global Sales by Platform')
plt.xlabel('Platform')
plt.ylabel('Total Global Sales (in millions)')
plt.show()
```

# 8. HOW GLOBAL SALES OF VIDEO GAMES HAVE EVOLVED OVER THE YEARS 


```python
a = df.groupby('Year')['Global_Sales'].sum()
print(a)

plt.figure(figsize=(10, 6))
plt.plot(a.index,a)
plt.title('Global Sales Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Global Sales (in millions)')
plt.show()
```

# 9. HIGHEST AVERAGE GLOBAL SALES BY GENRE


```python
s = df.groupby('Genre')['Global_Sales'].mean().sort_values(ascending=False)
print("Genre with the highest average global sales:")
print(s.head(1))
s.plot(kind='bar')
plt.title('Average Global Sales by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Global Sales (in millions)')
plt.show()
```

# 10. TOP 10 VIDEO GAMES BY GLOBAL SALES

```python
# 10. TOP 10 VIDEO GAMES BY GLOBAL SALES

a = df.sort_values(by='Global_Sales', ascending=False).head(10)
print("Top 10 video games by global sales:")
print(a[['Name', 'Global_Sales']])
plt.figure(figsize=(12, 6))
sns.barplot(data=a, x='Global_Sales', y='Name')
plt.title('Top 10 Video Games by Global Sales')
plt.xlabel('Global Sales (in millions)')
plt.ylabel('Video Game')
plt.show()
```

# 11. AVERAGE SALES BY EACH REGION


```python
a = df[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].mean().reset_index()
a.columns = ['Region', 'Average_Sales']
plt.figure(figsize=(12, 5))
sns.barplot(x='Region', y='Average_Sales', data=a)
plt.title('Average Sales by Region')
plt.xlabel('Region')
plt.ylabel('Average Sales (in millions)')
plt.show()
```

# 12. PIE CHART OF GENRE WISE LAUNCHING OF GAMES 

```python
plt.figure(figsize=(10, 6))
a=df['Genre'].value_counts()
a.plot(kind='pie',autopct='%1.1f%%')
plt.title("Genre Wise Launching of Games")
plt.ylabel('')  # Hide the y-label
plt.show()
```

# 13. NUMBER OF GAMES LAUNCHED PER YEAR 


```python
# 13 NUMBER OF GAMES LAUNCHED PER YEAR 

df = df.sort_values(by='Year')
plt.figure(figsize=(20, 8))
sns.countplot(x='Year', data=df)
plt.ylabel('Number of Games launched')
plt.show()
```
